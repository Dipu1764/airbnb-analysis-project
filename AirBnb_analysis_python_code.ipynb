```python
import pandas as pd

# Load listings data
chicago_df = pd.read_csv('/content/listings chicago.csv')
nola_df = pd.read_csv('/content/listings new Orleans.csv')

# Add city column to each
chicago_df['city'] = 'Chicago'
nola_df['city'] = 'New Orleans'

# Combine both datasets
df = pd.concat([chicago_df, nola_df], ignore_index=True)

# Select relevant columns
columns_to_keep = [
    'id', 'name', 'host_id', 'host_name', 'neighbourhood_group',
    'neighbourhood', 'latitude', 'longitude', 'room_type', 'price',
    'minimum_nights', 'number_of_reviews', 'last_review', 'reviews_per_month',
    'availability_365', 'city'
]
df = df[columns_to_keep]

# Clean price column (remove $ and commas, convert to float)
df['price'] = df['price'].replace('[\$,]', '', regex=True).astype(float)

# Handle missing values (optional: drop or fill)
df['reviews_per_month'] = df['reviews_per_month'].fillna(0)
df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')

# Export cleaned dataset
df.to_csv('airbnb_combined_cleaned.csv', index=False)
```
